##### This is a notebook that performs an EDA. It first loads the "250710-bp3d-db-data.csv" file, then displays the total item count in the ".csv" file, considering columnsXrow only. It also displays key data such as column composition, name, and quantity of the "units" to be burned. Finally, it analyzes the content of the columns that are of interest for Sentitivity Analysis (SA), but just on position [0].

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pprint import pprint
```

```python
#Charge or full "250710-bp3d-db-data" file (3.5 GB).
full_df = pd.read_csv("250710-bp3d-db-data.csv")
```

```python
# Total number of items
# Number of rows
print("Rows:", len(full_df))

# Number of columns
print("Columns:", len(full_df.columns))

# Total number of cells (items)
total_items = full_df.size
print("Total items:", total_items)
```

```python
# Let's look at the columns
full_df.columns
```

```python
# Now let's check the number of records per 'unit name' 
i = 0 
for unit, count in full_df["unit_name"].value_counts().items():
   i += 1
   print(f"{unit}: {count}")
# Depends on your IDE, this part can be shown as a scrollable object or be printed completely
```

```python
print("There are", i, "units in total.")
```

```python
# Keep only the columns that we need, based in Ismael and Pedro's analysis.
columns_to_keep = ['fuel_id',
                   'unit_feature_id',
                   'run_parameters',
                   'ignition_parameters',
                   'ignition_pattern',
                   'metrics', ]
sub_df = full_df[columns_to_keep]
```

```python
# You can access the first element (index 0) of specific columns in sub_df.
# Create a dictionary to hold the collected data for better organization and display.
# Each key in the dictionary corresponds to a column name, and its value is the data
# retrieved from the first row of that column.
display_data = {
    'run_parameters': sub_df['run_parameters'][0],
    'ignition_parameters': sub_df['ignition_parameters'][0],
    #'ignition_pattern': sub_df['ignition_pattern'][0], # These are just points, uncomment this line if you want to look at all of them
    'metrics': sub_df['metrics'][0]
}
pprint(display_data, width=200, depth=None)
```
